````python
def extract_text_from_pdf(pdf_path):
  text=""
  with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
      text+=page.extract_text()+"\n"
  return text
````

````python
import json
def extract_knowledge(text):
  prompt=f""" Extract entities and relationships from the following text.
  Return ONLY JSON in this format:
  {{
  "entities":[
    "Entity1",
    "Entity2"
  ],
  "relationships":[
  {{
    "source":"Entity1",
    "target":"Entity2",
    "relation":"relationship"
    }}
    ]
    }}
    Text:
    {text}
    """
  response=model.generate_content(prompt)
  json_text=response.text.strip()

  json_text=json_text.replace("```json","")
  json_text=json_text.replace("```","")
  return json.loads(json_text)
````

````python
# create knowledge graph
def create_graph(data):
  G=nx.DiGraph()
  for entity in data["entities"]:
    G.add_node(entity)
  for rel in data["relationships"]:
    G.add_edge(
        rel["source"],
        rel["target"],
        relation=rel["relation"]
    )
  return G
````

````python
def visualize_graph(G):
  plt.figure(figsize=(12,8))
  pos=nx.spring_layout(G,seed=42)
  nx.draw(
      G,
      pos,
      with_labels=True,
      node_size=2500,
      node_color="lightblue",
      font_size=10,
      arrows=True
  )
  edge_labels =nx.get_edge_attributes(G,'relation')
  nx.draw_networkx_edge_labels(
      G,
      pos,
      edge_labels=edge_labels
  )
  plt.title("Knowledge Graph")
  plt.show()
````

````python
pdf_path="/content/AI-Powered_Real-Time_Crowd_Behavioral_Anomaly_Detection_with_Alert_Sound_System.pdf"
document_text=extract_text_from_pdf(pdf_path)
document_text=document_text[:10000]
knowledge=extract_knowledge(document_text)
print(json.dumps(knowledge,indent=4))
graph=create_graph(knowledge)
visualize_graph(graph)
````

````python
def graph_to_text(graph):
  context=[]
  for source,target,data in graph.edges(data=True):
    relation=data.get(
        "relation"
        "relation_to"
    )
    context.append(
        f"{source} {relation} {target}"
    )
  return "\n".join(context)
````

````python
def ask_question(graph,question):
  graph_context=graph_to_text(graph)
  prompt=f"""
  you are a knowledge Graph Question Answering System.
  Knowledge Graph Facts:
  {graph_context}
  User Question:
  {question}

  Rules:
  1. Answer only from the graph facts.
  2. Do not invent information
  3. If answer is Unavailable, reply:
  "information not found in the knoweldge graph."
  """
  response=model.generate_content(prompt)
  return response.text
````

````python
def main():
  pdf_path="/content/AI-Powered_Real-Time_Crowd_Behavioral_Anomaly_Detection_with_Alert_Sound_System.pdf"
  document_text=extract_text_from_pdf(pdf_path)
  document_text=document_text[:10000]
  knowledge=extract_knowledge(document_text)
  print(json.dumps(knowledge,indent=4))
  graph=create_graph(knowledge)
  visualize_graph(graph)
  print("Knowledge Graph QA System")
  print("================================================")
  while True:
    question=input("\n Ask a question(type exit to quit):")
    if question.lower()=="exit":
      print("Exiting...")
      break
    answer =ask_question(
        graph,
        question
    )
    print("\n Answer")
    print(answer)
if __name__=="__main__":
  main()
````

## 9. Build a multiagent multi-step workflow orchestration demonstrating task sequencing, retry handling and failure management ensuring durability and reliability of the workflows for a suitable use case

````python
import google.generativeai as genai
import json
import time
from datetime import datetime

# Configure your Gemini API Key
genai.configure(api_key="AQ.Ab8RN6ITrDIPCG9cm1Ph4sm7SLNBvgMwjgn3WAjnUlIwo5Xv1w")

# Load Gemini Model
model = genai.GenerativeModel("gemini-2.5-flash")
````

````python
# Stores workflow progress
workflow_state = {}

# Save state to JSON file
def save_state():

    with open("workflow_state.json", "w") as file:

        json.dump(workflow_state, file, indent=4)

    print("Workflow state saved.")
````

````python
def ask_gemini(prompt):

    retries = 3

    for attempt in range(retries):

        try:

            print(f"\nGemini API Call (Attempt {attempt+1})")

            response = model.generate_content(prompt)

            return response.text

        except Exception as e:

            print("Error :", e)

            if attempt < retries-1:

                print("Retrying in 2 seconds...")

                time.sleep(2)

            else:

                print("Maximum retries reached.")

                return "FAILED"
````

````python
def weather_agent(city):

    print("\nWeather Agent Started")

    prompt = f"""
    You are a weather forecasting assistant.

    Give today's weather forecast for {city}.

    Include:
    - Temperature
    - Humidity
    - Weather Condition
    """

    weather = ask_gemini(prompt)

    workflow_state["Weather"] = weather

    save_state()

    return weather
````

````python
def clothing_agent(weather):

    print("\nClothing Recommendation Agent Started")

    prompt = f"""
    Based on the following weather:

    {weather}

    Suggest suitable clothing.
    """

    clothes = ask_gemini(prompt)

    workflow_state["Clothing"] = clothes

    save_state()

    return clothes
````

````python
def travel_agent(weather):

    print("\nTravel Advisory Agent Started")

    prompt = f"""
    Based on the following weather:

    {weather}

    Give travel advice.

    Mention:

    - Whether travelling is safe
    - Things to carry
    - Any precautions
    """

    travel = ask_gemini(prompt)

    workflow_state["Travel"] = travel

    save_state()

    return travel
````

````python
def emergency_agent(weather):

    print("\nEmergency Alert Agent Started")

    prompt = f"""
    Analyze the following weather.

    {weather}

    If there is any severe weather like heavy rain,
    storm, cyclone or flood, provide an emergency alert.

    Otherwise say:
    "No Emergency Alert."
    """

    alert = ask_gemini(prompt)

    workflow_state["Emergency Alert"] = alert

    save_state()

    return alert
````

````python
def report_agent():

    print("\nReport Generator Agent Started")

    prompt = f"""
    Generate a final weather report.

    Weather Information:
    {workflow_state["Weather"]}

    Clothing Recommendation:
    {workflow_state["Clothing"]}

    Travel Advice:
    {workflow_state["Travel"]}

    Emergency Alert:
    {workflow_state["Emergency Alert"]}

    Prepare a neat report.
    """

    report = ask_gemini(prompt)

    workflow_state["Final Report"] = report

    save_state()

    return report
````

````python
def weather_workflow(city):

    print("\n===================================")
    print(" WEATHER WORKFLOW STARTED ")
    print("===================================")

    # ----------------------------
    # Step 1
    # ----------------------------

    weather = weather_agent(city)

    if weather == "FAILED":

        print("Weather Agent Failed")
        print("Workflow Terminated")

        return

    # ----------------------------
    # Step 2
    # ----------------------------

    clothes = clothing_agent(weather)

    if clothes == "FAILED":

        print("Clothing Agent Failed")
        print("Workflow Terminated")

        return

    # ----------------------------
    # Step 3
    # ----------------------------

    travel = travel_agent(weather)

    if travel == "FAILED":

        print("Travel Agent Failed")
        print("Workflow Terminated")

        return

    # ----------------------------
    # Step 4
    # ----------------------------

    emergency = emergency_agent(weather)

    if emergency == "FAILED":

        print("Emergency Agent Failed")
        print("Workflow Terminated")

        return

    # ----------------------------
    # Step 5
    # ----------------------------

    report = report_agent()

    if report == "FAILED":

        print("Report Generation Failed")
        print("Workflow Terminated")

        return

    print("\n===================================")
    print(" WORKFLOW COMPLETED ")
    print("===================================")

    print("\nFINAL REPORT\n")

    print(report)
````
